```python
import pandas as pd

df = pd.read_excel("../raw/Superstore.xls", sheet_name="Orders")
print(df.shape)
print(df.columns.tolist())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print(
    "Profit Margin %:",
    round(total_profit / total_sales * 100, 2)
)
```

```python
loss = df[df["Profit"] < 0]

print("Loss-making rows:", len(loss))
print("Loss-making %:", round(len(loss) / len(df) * 100, 2))
print("Total loss:", round(loss["Profit"].sum(), 2))
```

```python
temp = df.copy()

temp["Year"] = pd.to_datetime(temp["Order Date"]).dt.year

yearly = temp.groupby("Year").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).reset_index()

yearly["Margin %"] = (
    yearly["Profit"] / yearly["Sales"] * 100
).round(2)

yearly["Sales YoY %"] = (
    yearly["Sales"].pct_change() * 100
).round(2)

yearly["Profit YoY %"] = (
    yearly["Profit"].pct_change() * 100
).round(2)

yearly
```

```python
temp["Discount Band"] = pd.cut(
    temp["Discount"],
    bins=[-0.01, 0, 0.10, 0.20, 0.30, 0.40, 1.00],
    labels=["0%", "1-10%", "11-20%", "21-30%", "31-40%", ">40%"]
)
```

```python
temp["Discount Band"] = pd.cut(
    temp["Discount"],
    bins=[-0.01, 0, 0.10, 0.20, 0.30, 0.40, 1.00],
    labels=["0%", "1-10%", "11-20%", "21-30%", "31-40%", ">40%"]
)

band = temp.groupby("Discount Band", observed=True).agg(
    Rows=("Row ID", "count"),
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Avg_Discount=("Discount", "mean")
).reset_index()

band["Profit Margin %"] = (
    band["Profit"] / band["Sales"] * 100
).round(2)

band
```

```python
cat = df.groupby("Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Orders=("Order ID", "nunique"),
    Avg_Discount=("Discount", "mean")
).reset_index()

cat["Profit Margin %"] = (
    cat["Profit"] / cat["Sales"] * 100
).round(2)

cat.sort_values("Profit")
```

```python
print("Fully duplicated rows:", df.duplicated().sum())
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())
print("Unique Order IDs:", df["Order ID"].nunique())
print("Ship before order:", (df["Ship Date"] < df["Order Date"]).sum())
print("Unique sub-categories:", df["Sub-Category"].nunique())
print("Unique states:", df["State/Province"].nunique())
print("Unique customers:", df["Customer ID"].nunique())
```

```python
print(df["Country/Region"].value_counts())
print()
print(sorted(df["State/Province"].unique()))
```

```python
text_cols = df.select_dtypes(include=["object", "string"]).columns
for col in text_cols:
    s = df[col].astype(str)
    spaces = (s != s.str.strip()).sum()
    print(f"{col}: {spaces} values with leading/trailing spaces")
```

```python
print(df["Postal Code"].map(type).value_counts())
print(df["Postal Code"].astype(str).str.len().value_counts())
```

```python
chk = df.groupby("Order ID").agg(
    customers=("Customer ID", "nunique"),
    order_dates=("Order Date", "nunique")
)
print("Orders with >1 customer:", (chk["customers"] > 1).sum())
print("Orders with >1 order date:", (chk["order_dates"] > 1).sum())
```

```python
names_per_id = df.groupby("Product ID")["Product Name"].nunique()
print("Product IDs with >1 name:", (names_per_id > 1).sum())
```

```python
cols = ["Order ID", "Sub-Category", "Product Name", "Sales", "Quantity", "Discount", "Profit"]
print("Top 5 profit rows")
display(df.nlargest(5, "Profit")[cols])
print("Bottom 5 profit rows")
display(df.nsmallest(5, "Profit")[cols])
```

```python
print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
print("Profit greater than Sales:", (df["Profit"] > df["Sales"]).sum())
print("Zero-profit rows:", (df["Profit"] == 0).sum())
```

```python
canada = df[df["Country/Region"] == "Canada"]

print(canada["Region"].value_counts())

print(
    "Sales:", round(canada["Sales"].sum(), 2),
    "| Profit:", round(canada["Profit"].sum(), 2)
)

print(
    "Canada margin %:",
    round(canada["Profit"].sum() / canada["Sales"].sum() * 100, 2)
)

display(
    canada[["City", "State/Province", "Postal Code", "Region"]].head()
)
```
